### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

1. What are some reasons that animals might migrate?

As was discussed in class, animals migrate for a variety of reasons including following food resources, warmer temperatures, or for breeding purposes (National Geographic, 2024). For example, in the Serengeti in East Africa, wildebeests as well as zebras and gazelles migrate looking for fresh grass which can be hard to find during the dry season (Trotter, 2024). As an aside, in college I was lucky to go to Tanzania and witness the wildebeest migration. We were tasked with completing point counts of animal sightings. While some groups reported back a few elephants or small packs of hyenas, my group was lucky enough to see hundreds upon hundreds of ungulates migration across the plains. It was magnificent!

Monarch butterflies are an example of a species that migrates seasonally. The Eastern Monarch migrates throughout North America, spending summers along the East Coast of the United States and as far north as Southern Canada and then journeying to Central Mexico for winter once shorter days, cooler temperatures, and limited food sources drive them south. One thing that is fascinating about Monarchs is that their migration spans multiple generations of butterflies. The first-generation travels North from Central Mexico, second and third generations summer in the US and Canada, and the fourth-generation travels back South to Central Mexico (Migration). It is amazing to wonder how they know where to go and it is a life-long goal of mine to travel to Mexico so see the UNESCO World Heritage site where millions of Monarchs congregate.

2. How might climate change affect animal migrations?

Climate change is having a significant impact on movement and migration of wildlife globally. These impacts include loss of habitat and changes in timing and extent of migratory patterns (Martay et al, 2023). For example, Monarch Butterflies are facing a number of threats including loss of habitat, loss of food sources due to drought, more treacherous migration due to extreme weather, and we are seeing a shift in timing of migration likely due to warming temperatures (Neff, 2021). Additionally, the Wildebeest migration has become more irregular with animals spending less time in certain areas due to drought and limited food sources (Kerongo Mosa, 2024). 

3. Do you notice any animal migrations where you live?

In Durango, we notice the migratory patterns of both Elk and Canadian Geese among many species. This time of year, we see the Elk move down from the mountains closer to town for the winter for more food and warmer weather (Wildlife Migration and Movement, 2023). Anecdotally, this year was very dry, and we saw Elk in and around town earlier in the fall than most years. Canadian Geese can be seen flying overhead in spring and fall as they move North and then South. The Geese have also had to shift their migratory patterns due to climate change, both delaying their migration and not traveling as far South as they have historically (Hooper, 2020).
